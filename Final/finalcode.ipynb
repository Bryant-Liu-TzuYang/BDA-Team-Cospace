```python
import requests
import json
import time
import re
import csv
import numpy as np
import pandas as pd

API_KEY = "Your API key"
url = "https://api.perplexity.ai/chat/completions"

# 確保他回傳的json合法
def safe_json_loads(content):
    try:
        return json.loads(content)
    except:
        match = re.search(r'\[.*\]', content, re.DOTALL)
        if match:
            try:
                return json.loads(match.group())
            except:
                pass
        return []

# 主要調用LLM的函數
def get_keywords_batch(texts):
    combined_text = "\n".join(texts)
    payload = {
        "model": "llama-3.1-sonar-large-128k-online",
        "messages": [
            {
                "role": "system",
                "content": (
                    "你是一個關鍵字提取助手，請從用戶提供的每個商品名稱及敘述中，"
                    "為每個商品分別提取三個關鍵字詞，每個關鍵字詞長度2到10字，且不能是全數字。"
                    "請**只**以JSON格式回傳所有商品的關鍵字詞列表，格式為："
                    "[\"商品1關鍵字1\", \"商品1關鍵字2\", \"商品1關鍵字3\", \"商品2關鍵字1\", ...]。"
                    "不要回傳任何說明、前後標註或多餘內容。"
                )
            },
            {
                "role": "user",
                "content": f"請從以下文字提取關鍵字詞：\n{combined_text}"
            }
        ],
        "temperature": 0.2,
        "top_p": 0.9,
        "return_citations": False
    }
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }
    response = requests.post(url, json=payload, headers=headers)
    if response.status_code == 200:
        try:
            data = response.json()
            content = data["choices"][0]["message"]["content"]
            # print("API回傳內容：", content[:200])  # 只印前200字方便檢查
            keywords = safe_json_loads(content)
            return keywords
        except Exception as e:
            print(f"解析錯誤: {e}")
            return []
    else:
        print(f"API錯誤: {response.status_code} {response.text}")
        return []

# 弄出純文字敘述的列表
def combine_title_description(df):
    return [f"{row['SalePageTitle']} {row['SaleProductDescShortContent']}" for _, row in df.iterrows()]


# 主程式區域 -----------

monthrange = ['01','02','03','04','05','06','07','08','09','10','11','12']
all_result = []

for i in monthrange:
    filepath = 'gmv1000bymonth/month' +i+ '.csv'
    print(filepath)

    monthdata = pd.read_csv(filepath)
    #print(monthdata.head())

    puredisc = combine_title_description(monthdata)
    #print(puredisc[1])

    # 批次處理，每批50個商品（建議先測試50個）
    all_keywords = []
    batch_size = 50
    for i in range(0, len(puredisc), batch_size):
        batch = puredisc[i:i+batch_size]
        keywords = get_keywords_batch(batch)
        all_keywords.extend(keywords)
        time.sleep(0.5)

    all_result.append(all_keywords)

print(all_result[0])
print(all_result[-1])
```

```python
import pandas as pd
import numpy as np

data1 = all_result  # 包含 12 個子列表的列表

# 步驟 1：檢查數據結構
print("all_result 類型:", type(data1))
print("子列表數量:", len(data1))
for i, sublist in enumerate(data1):
    print(f"Month_{i+1} 類型: {type(sublist)}, 長度: {len(sublist)}")
    # print(f"Month_{i+1} 前3個元素: {sublist[:3] if len(sublist) >= 3 else sublist}")

# 步驟 2：檢查每個子列表的長度
lengths = [len(sublist) for sublist in data1]
max_length = max(lengths)
print("每個子列表的長度:", lengths)
print("最大長度（CSV 行數）:", max_length)

# 步驟 3：將所有子列表填充到相同長度
padded_data = [sublist + [np.nan] * (max_length - len(sublist)) for sublist in data1]

# 步驟 4：創建 DataFrame
try:
    df = pd.DataFrame({f"Month_{i+1}": sublist for i, sublist in enumerate(padded_data)})
except ValueError as e:
    print("創建 DataFrame 時出錯:", e)
    raise

# 步驟 5：將 DataFrame 儲存為 CSV，保留索引
df.to_csv('keywords_of_12mon.csv', index=True, encoding='utf-8')

# 步驟 6：打印前幾行以驗證
print("前 5 行數據預覽：")
print(df.head())
print("總行數:", len(df))
```
